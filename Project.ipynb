In recent years, the Great Lakes have experienced an increase in flooding, believed to be caused by rapid environmental changes from global warming onset. In this analysis, I will be reviewing the last 100 years of data on the water levels of each lake in meters (IGLD85), which represents "the height equivalent (in meters) of the work needed to raise a unit mass of water from mean sea level to the point of measurement." To achieve this, I will be calculating various summary statistics as well as producing multiple graphs to visualize how each lake's water levels have changed within the last century.

To start, I will import the two packages I will be using in this analysis for math and plotting.

```python
import numpy as np
import matplotlib.pyplot as plt
```

I will now load my datasets into select variables (data from NOAA).

```python
mh_year, mh_level = np.loadtxt("lake_michigan_huron.csv", usecols = [0,1], unpack=True, delimiter=',')
erie_year, erie_level = np.loadtxt("lake_erie.csv", usecols = [0,1], unpack=True, delimiter=',')
superior_year, superior_level = np.loadtxt("lake_superior.csv", usecols = [0,1], unpack=True, delimiter=',')
ontario_year, ontario_level = np.loadtxt("lake_ontario.csv", usecols = [0,1], unpack=True, delimiter=',')
```

### Guiding Questions:

*Which lake has the highest water level historically?*

*Which lake has the most variance in water level?*

*Was there any correlation between the lakes?*

*Could current global warming be a cause of the recent changes?*

### Part 1: Descriptive Statistics

In this section I will use NumPy to calculate some basic statistics to help understand the basics of each dataset

```python
print(f"Michigan/Huron Mean: {np.mean(mh_level)}")
print(f"Erie Mean: {np.mean(erie_level)}")
print(f"Superior Mean: {np.mean(superior_level)}")
print(f"Ontario Mean: {np.mean(ontario_level)}")
```

Based on the means, we can see that over the datasets' interval, Lake Superior has, on average, the highest water level out of the three. While this is a nice statistic to know, we cannot infer much about the changes that have taken place.

```python
print(f"Michigan/Huron Standard Dev:{np.std(mh_level)}")
print(f"Erie Standard Dev:{np.std(erie_level)}")
print(f"Superior Standard Dev:{np.std(superior_level)}")
print(f"Ontario Standard Dev:{np.std(ontario_level)}")
```

Now that we have more information, we can make a better conclusion. Although Superior had the highest mean, it has the lowest amount of variation, meaning most likely it has had the least amount of growth overtime. Michigan/Huron had the highest variation. To fully compare growth, I will visualize plots in the next part of the analysis. 

Before moving on I will calculate the correlation coefficient between the years and each lakes level to estimate the relationship between them. 

```python
print(f"Year/Michigan/Huron:{np.corrcoef(mh_year,mh_level)}")
print(f"Year/Erie:{np.corrcoef(erie_year,erie_level)}")
print(f"Year/Superior:{np.corrcoef(superior_year,superior_level)}")
print(f"Year/Ontario:{np.corrcoef(ontario_year,ontario_level)}")
```

Looking at these values, Erie has the highest correlation at .49 meaning it has a moderate positive correlation while Superior has a negative correlation value of -.0099 meaning it has a weak negative correlation. This suggests the notion that Superiors high mean meant a higher level of change over time. 

### Part 2: Visualizing Data 

In this section of the analysis, I will be using the data sets to create multiple plots to gain a better understanding of each dataset as well as seeing how they interact with each other. 

This first group of graphs will visualize each dataset in their own plots with standardized axes to compare the changes to each other. 

```python
plt.figure(figsize = (40,20))
plt.suptitle("Great Lakes Water Level (Meters) x Time (Years)",fontsize = 50)

plt.subplot(2,2,1)
plt.plot(mh_year,mh_level, color = "blue",marker = "o",ms=3)
plt.title("Michigan/Huron", fontsize = 25)
plt.xlabel("Time (Years)", fontsize = 20)
plt.ylabel("Water Level (Meters)",fontsize = 20)
plt.xlim(1950,2021)

plt.subplot(2,2,2)
plt.plot(erie_year,erie_level, color = "purple",marker = "o",ms=3)
plt.title("Erie", fontsize = 25)
plt.xlabel("Time (Years)", fontsize = 20)
plt.ylabel("Water Level (Meters)",fontsize = 20)
plt.xlim(1950,2021)

plt.subplot(2,2,3)
plt.plot(superior_year,superior_level,color = "green",marker = "o",ms=3)
plt.title("Superior", fontsize = 25)
plt.xlabel("Time (Years)", fontsize = 20)
plt.ylabel("Water Level (Meters)",fontsize = 20)
plt.xlim(1950,2021)

plt.subplot(2,2,4)
plt.plot(ontario_year,ontario_level,color = "orange",marker = "o",ms=3)
plt.title("Ontario", fontsize = 25)
plt.xlabel("Time (Years)", fontsize = 20)
plt.ylabel("Water Level (Meters)",fontsize = 20)
plt.xlim(1950,2021)

```

#### Summary:

The plots ahead show the changes in water level as time increases across the interval 1950-2021. I choose to shorten the time frame to make the data easier to display without becoming too crowded. This also allows us to view more recent history. When looking at these graphs, there are a few patterns that stand out. All four charts are very volatile and feature many hard up and down changes in water level. While this does agree with the small standard deviations and weaker correlation, a big part of this may be due to the small y-scale that all the data is included in. The most important observation is towards the end of the graph. Each one has a dip around 2000 - 2015 followed by spikes up to 2021. While this isn't abnormal, as there are many surges seen in previous years, if this trend continues, we may be able to explain this change with other environmental changes surrounding global warming. These feelings are also reinforced by the fact that the peaks of the most recent years are either the highest peaks or on par with the highest peaks of the past. 

This next chart will have all 4 datasets plotted on one graph so that I can more efficiently compare the different water levels. 

```python
plt.figure(figsize=(15,7))
plt.plot(mh_year,mh_level, color = "blue",label="Michigan/Huron")
plt.plot(erie_year,erie_level, color = "purple",label="Erie")
plt.plot(superior_year,superior_level,color = "green",label="Superior")
plt.plot(ontario_year,ontario_level,color = "orange",label="Ontario")
plt.ylim(173,184)
plt.title("Great Lakes Water Level Comparison")
plt.ylabel("Water Level (Meters)")
plt.xlabel("Time (Years)")
plt.grid()
plt.legend()
```

This graph shows Erie, Michigan/Huron, and Superior only because Ontario had to low of a mean to effectively be displayed with the rest. Now that we can see them together in one plot, we can confirm that they indeed do follow a similar pattern over the course of the century. The variability is still consistent throughout all the datasets as well as the increase starting around 2010. You can also see that the local mins and maxes of each dataset do seem to be correlated. I will explore this concept more in the next section. 

The graphs below will plot each lakes dataset against the other lakes in a scatterplot to identify correlation and spot patterns in the level increases. 

```python
plt.figure(figsize = (20,20))
plt.suptitle("Great Lakes Water Level Correlation",fontsize = 35)

plt.subplot(2,3,1)
plt.scatter(mh_level,ontario_level,color ="blue")
plt.xlabel("Michigan/Huron Water Level (Meters)",fontsize = 15)
plt.ylabel("Ontario Water Level (Meters)",fontsize = 15)

plt.subplot(2,3,2)
plt.scatter(mh_level,superior_level,color ="purple")
plt.xlabel("Michigan/Huron Water Level (Meters)",fontsize = 15)
plt.ylabel("Superior Water Level (Meters)",fontsize = 15)

plt.subplot(2,3,3)
plt.scatter(mh_level,erie_level,color ="green")
plt.xlabel("Michigan/Huron Water Level (Meters)",fontsize = 15)
plt.ylabel("Erie Water Level (Meters)",fontsize = 15)


plt.subplot(2,3,4)
plt.scatter(ontario_level,superior_level,color ="yellow")
plt.xlabel("Ontario Water Level (Meters)",fontsize = 15)
plt.ylabel("Superior Water Level (Meters)",fontsize = 15)


plt.subplot(2,3,5)
plt.scatter(ontario_level,erie_level,color ="orange")
plt.xlabel("Ontario Water Level (Meters)",fontsize = 15)
plt.ylabel("Erie Water Level (Meters)",fontsize = 15)

plt.subplot(2,3,6)
plt.scatter(superior_level,erie_level,color ="black")
plt.xlabel("Superior Level (Meters)",fontsize = 15)
plt.ylabel("Erie Water Level (Meters)",fontsize = 15)
```

This figure proves that there is some form of correlation between increased water levels between the lakes. Although all six graphs have different appearances, they all have a positive correlation. What this most likely means is that some outside factors influence all the lakes mins and maxes and cause these increases and decreases. The strongest correlation can be found between Erie, Michigan/Huron, and Ontario, meaning that a factor or location affects these lakes heavier than the other two. The weakest correlation is between Ontario and Superior, meaning that these two most likely have the most different confounding factors.
